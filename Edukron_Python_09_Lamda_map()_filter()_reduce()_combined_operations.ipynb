```python
### 20. Using reduce(), calculate the sum of unique numbers only. Do not directly use sum(set(numbers)).

numbers = [12, 45, 12, 67, 45, 89, 23, 67, 100]

# The accumulator `acc` will be a tuple: (current_sum, set_of_unique_numbers_seen)
# Initial value for reduce is (0, set()) to store the initial sum and an empty set.
result = reduce(
    lambda acc, y: (acc[0] + y, acc[1] | {y}) if y not in acc[1] else acc, numbers, (0, set())
)

# The reduce function returns the final tuple, so we extract the sum.
unique_sum = result[0]

print("Sum of unique numbers: ", unique_sum)
```

### === Section 5: Combined Lambda, map(), filter(), and reduce() — 20 Questions

```python
### 1. Using filter() and map():

# Extract all even numbers.
# Calculate the square of every even number.
# Return the squared values as a list.

numbers = [12, 45, 67, 89, 23, 56, 91, 34, 78, 100]

even_numbers = filter(lambda x : x % 2 == 0, numbers)
## x % 2 == 0 returns even
## list() converts the result into list
squared_even = list(map(lambda x : x ** 2, even_numbers))

print(squared_even)
```

```python
### 2. Using filter() and reduce():

# Extract all odd numbers.
# Calculate their total sum.


numbers = [12, 45, 67, 89, 23, 56, 91, 34, 78, 100]

odd_numbers = filter(lambda x : x % 2 != 0, numbers)
## x != 0 returns odd
total_sum = reduce(lambda x, y : x + y, odd_numbers)

print(total_sum)
```

```python
### 3. Using filter() and map():

# Extract numbers greater than 50.
# Multiply every selected number by 2.

numbers = [10, 25, 40, 55, 70, 85, 100]

# filter numbers > 50
filtered_numbers = filter(lambda x : x > 50, numbers)

# Multiply each filtered number by 2
result = list(map(lambda x : x * 2, filtered_numbers))

print(result)
```

```python
### 4.Using filter(), map(), and reduce():

# Extract positive numbers.
# Calculate the square of every positive number.
# Calculate the total sum of the squared values.

numbers = [-15, 20, -35, 40, 50, -10, 60]
 ## x > 0 returns positive numbers
positive_numbers = filter(lambda x : x > 0, numbers)
 ## x ** 2 returns the square of positive numbers
squared_positive_numbers = map(lambda x : x ** 2, positive_numbers)
 ## reduce calculates the sum of squared positive numbers
total_sum = reduce(lambda x, y : x + y, squared_positive_numbers)

print(total_sum)
```

```python
### 5. Using filter() and map():

# Extract words containing more than five characters.
# Convert the selected words to uppercase.

words = ["python", "azure", "ai", "developer", "cloud", "analytics", "spark"]

filtered_words = filter(lambda x : len(x) > 5, words)
uppercase_words = list(map(lambda x : x.upper(), filtered_words))

print(uppercase_words)
```

```python
### 6. Using filter() and map():

# Extract names beginning with "A".
# Convert the selected names to lowercase.

names = ["Anil", "Rahul", "Ajay", "Meena", "Arjun", "Kiran", "Akhil"]

filtered_names = filter(lambda x : x.startswith("A"), names)
## startswith("A") returns value of name starting with A
## x.lowercase() , filtered_names return the names in lowercase
lower_case = list(map(lambda x : x.lower(), filtered_names))

print(lower_case)
```

```python
### 7. Using filter() and reduce():

# Extract numbers divisible by both 3 and 5.
# Calculate their total sum.

numbers = [3, 5, 10, 15, 18, 30, 42, 45, 60, 75]

divisible_by_3_5 = filter(lambda x : x % 3 == 0 and x % 5 == 0, numbers)
total_sum = reduce(lambda x, y : x + y, divisible_by_3_5)

print("Total Sum: ", total_sum)

```

```python
### 8. Using filter() and reduce():

# Extract palindrome words.
# Join them into one string separated by " | ".

words = ["madam", "python", "level", "azure", "radar", "cloud", "civic"]
## original string is equal to reverse string is Palindrome String
# x == x [::-1] returns the palindrome value
palindrome_words = filter(lambda x : x == x[::-1], words)
## x + " | " + y joins string & seperated by " | "
joined_string = reduce(lambda x, y : x + " | " + y, palindrome_words)

print("Joined string: ", joined_string)
```

```python
### 9. Using filter(), map(), and reduce():

# Extract temperatures greater than 30°C.
# Convert the selected values into Fahrenheit.
# Calculate the average Fahrenheit temperature.

temperatures = [18, 24, 30, 35, 40, 12, 28, 33, 15, 42]
## temperatures > than 30C
high_temp = filter(lambda x : x > 30, temperatures)
## Fahrenheit = (Celsius * 9/5) + 32 converts F to C
fahrenheit = list(map(lambda x : (x * 9/5) + 32, high_temp))

total = reduce(lambda x, y : x + y, fahrenheit)
average = total / len(fahrenheit)

print("Temperatures greater than 30°C: ", fahrenheit)
print("Average Fahrenheit temperature: ", average)
```

```python
### 13. Using filter(), map(), and reduce():

# Extract active employees.
# Increase every active employee's salary by 15%.
# Calculate the total updated payroll.

employees = [
    {"id": 101, "name": "Rahul", "salary": 45000, "active": True},
    {"id": 102, "name": "Anil", "salary": 65000, "active": True},
    {"id": 103, "name": "Sneha", "salary": 72000, "active": False},
    {"id": 104, "name": "Kiran", "salary": 82000, "active": True},
    {"id": 105, "name": "Meena", "salary": 55000, "active": True},
    {"id": 106, "name": "Arjun", "salary": 98000, "active": False}
]

# Step 1: Extract active employees
active_employees = list(filter(lambda emp: emp["active"] == True, employees))

# Step 2: Increase salary of active employees by 15%
updated_employees = list(map(
    lambda emp: {**emp, "salary": emp["salary"] * 1.15},
    active_employees
))

# Step 3: Calculate total updated payroll
total_payroll = reduce(
    lambda total, emp: total + emp["salary"],
    updated_employees,
    0
)

# Display results
print("Updated Active Employees:")
for emp in updated_employees:
    print(emp)

print("\nTotal Updated Payroll:", total_payroll)
```

```python
## 14. Using filter() and reduce():

# Extract students who passed.
# Find the topper among the passed students.

students = [
    {"name": "Ravi", "marks": 92, "passed": True},
    {"name": "Kiran", "marks": 68, "passed": True},
    {"name": "Meena", "marks": 85, "passed": True},
    {"name": "Ajay", "marks": 48, "passed": True},
    {"name": "Sneha", "marks": 35, "passed": False},
    {"name": "Rahul", "marks": 95, "passed": True}
]

# Step 1: Extract students who passed
passed_students = list(filter(lambda student: student["passed"] == True, students))

# Step 2: Find the topper among the passed students
topper = reduce(
    lambda s1, s2: s1 if s1["marks"] > s2["marks"] else s2,
    passed_students
)

# Display results
print("Passed Students:")
for student in passed_students:
    print(student)

print("\nTopper:")
print(topper)
```

```python
###  15. Using filter(), map(), and reduce():

# Extract students scoring more than 70.
# Extract only their marks.
# Calculate the average marks.

students = [
    {"name": "Ravi", "marks": 92},
    {"name": "Kiran", "marks": 68},
    {"name": "Meena", "marks": 85},
    {"name": "Ajay", "marks": 48},
    {"name": "Sneha", "marks": 79},
    {"name": "Rahul", "marks": 95}
]

# Step 1: Extract students scoring more than 70
filtered_students = list(filter(lambda student: student["marks"] > 70, students))

# Step 2: Extract only their marks
marks = list(map(lambda student: student["marks"], filtered_students))

# Step 3: Calculate the average marks
total_marks = reduce(lambda total, mark: total + mark, marks)
average_marks = total_marks / len(marks)

# Display results
print("Students scoring more than 70:")
print(filtered_students)

print("\nMarks:")
print(marks)

print("\nAverage Marks:", average_marks)
```

```python
### 16. Using filter(), map(), and reduce():

# Extract products currently in stock.
# Calculate each product's inventory value using price × stock.
# Calculate the total inventory value.

products = [
    {"name": "Laptop", "price": 55000, "stock": 5},
    {"name": "Mouse", "price": 600, "stock": 25},
    {"name": "Keyboard", "price": 1500, "stock": 0},
    {"name": "Monitor", "price": 18000, "stock": 8},
    {"name": "Printer", "price": 12000, "stock": 0},
    {"name": "SSD", "price": 6500, "stock": 12}
]

# Step 1: Extract products currently in stock
in_stock = list(filter(lambda product: product["stock"] > 0, products))

# Step 2: Calculate each product's inventory value (price × stock)
inventory = list(map(
    lambda product: {
        "name": product["name"],
        "inventory_value": product["price"] * product["stock"]
    },
    in_stock
))

# Step 3: Calculate the total inventory value
total_inventory_value = reduce(
    lambda total, product: total + product["inventory_value"],
    inventory,
    0
)

# Display results
print("Products in Stock:")
for product in in_stock:
    print(product)

print("\nInventory Value of Each Product:")
for item in inventory:
    print(item)

print("\nTotal Inventory Value:", total_inventory_value)
```

```python
### 17. Using filter(), map(), and reduce():

# Extract products belonging to the Electronics category.
# Apply a 20% discount.
# Calculate the total discounted value.

products = [
    {"name": "Laptop", "price": 55000, "category": "Electronics"},
    {"name": "Chair", "price": 4500, "category": "Furniture"},
    {"name": "Monitor", "price": 18000, "category": "Electronics"},
    {"name": "Desk", "price": 8500, "category": "Furniture"},
    {"name": "SSD", "price": 6500, "category": "Electronics"}
]

# Step 1: Extract Electronics products
electronics = list(filter(lambda product: product["category"] == "Electronics", products))

# Step 2: Apply a 20% discount
discounted_products = list(map(
    lambda product: {
        "name": product["name"],
        "discounted_price": product["price"] * 0.80
    },
    electronics
))

# Step 3: Calculate the total discounted value
total_discounted_value = reduce(
    lambda total, product: total + product["discounted_price"],
    discounted_products,
    0
)

# Display results
print("Electronics Products:")
for product in electronics:
    print(product)

print("\nDiscounted Products:")
for product in discounted_products:
    print(product)

print("\nTotal Discounted Value:", total_discounted_value)
```

```python
### 19. Using filter(), map(), and reduce():

# Extract transactions whose amount is greater than 5000.
# Add 18% GST to every selected transaction.
# Calculate the grand total after GST.

transactions = [
    {"id": 1, "amount": 1200},
    {"id": 2, "amount": 5200},
    {"id": 3, "amount": 18000},
    {"id": 4, "amount": 2500},
    {"id": 5, "amount": 9600},
    {"id": 6, "amount": 7200}
]

# Step 1: Extract transactions with amount greater than 5000
filtered_transactions = list(filter(lambda t: t["amount"] > 5000, transactions))

# Step 2: Add 18% GST to each selected transaction
gst_transactions = list(map(
    lambda t: {
        "id": t["id"],
        "amount_with_gst": t["amount"] * 1.18
    },
    filtered_transactions
))

# Step 3: Calculate the grand total after GST
grand_total = reduce(
    lambda total, t: total + t["amount_with_gst"],
    gst_transactions,
    0
)

# Display results
print("Transactions greater than 5000:")
for t in filtered_transactions:
    print(t)

print("\nTransactions after adding 18% GST:")
for t in gst_transactions:
    print(t)

print("\nGrand Total after GST:", grand_total)
```

```python
### 20. Using filter(), map(), and reduce():

# Extract active salespeople from the South region.
# Calculate a 5% incentive for each selected salesperson.
# Create a list containing salesperson name, sales amount, and incentive.
# Calculate the total incentive payable.


sales_records = [
    {"salesperson": "Rahul", "region": "South", "sales": 45000, "active": True},
    {"salesperson": "Anil", "region": "North", "sales": 65000, "active": True},
    {"salesperson": "Sneha", "region": "South", "sales": 72000, "active": False},
    {"salesperson": "Kiran", "region": "South", "sales": 82000, "active": True},
    {"salesperson": "Meena", "region": "West", "sales": 55000, "active": True},
    {"salesperson": "Arjun", "region": "South", "sales": 98000, "active": True}
]

# Step 1: Extract active salespeople from the South region
selected_salespeople = list(filter(
    lambda s: s["region"] == "South" and s["active"],
    sales_records
))

# Step 2 & 3: Calculate 5% incentive and create required list
incentives = list(map(
    lambda s: {
        "salesperson": s["salesperson"],
        "sales": s["sales"],
        "incentive": s["sales"] * 0.05
    },
    selected_salespeople
))

# Step 4: Calculate the total incentive payable
total_incentive = reduce(
    lambda total, s: total + s["incentive"],
    incentives,
    0
)

# Display results
print("Active Salespeople from South Region:")
for s in selected_salespeople:
    print(s)

print("\nSales and Incentives:")
for s in incentives:
    print(s)

print("\nTotal Incentive Payable:", total_incentive)
```
