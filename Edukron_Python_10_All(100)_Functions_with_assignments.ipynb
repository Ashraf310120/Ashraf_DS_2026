```python
# Assignment 12: Check Leap Year
# Write a function named is_leap_year().

# Function definition
# This function checks whether a given year is a leap year.
def is_leap_year(year):
    # Check the leap year conditions
    if year % 400 == 0:
        return True
    elif year % 4 == 0 and year % 100 != 0:
        return True
    else:
        return False


# Accept the year from the user
year = int(input("Enter a year: "))

# Call the function and store the result
result = is_leap_year(year)

# Display the appropriate message
if result:
    print(year, "is a Leap Year.")
else:
    print(year, "is not a Leap Year.")
```

```python
# Assignment 13: Factorial Using a Function
# Write a function named calculate_factorial().

# Function definition
# This function calculates the factorial of a non-negative integer.
def calculate_factorial(number):
    # Check for negative input
    if number < 0:
        return "Error! Factorial is not defined for negative numbers."

    # Initialize factorial to 1
    factorial = 1

    # Calculate the factorial using a loop
    for i in range(1, number + 1):
        factorial *= i

    # Return the factorial
    return factorial


# Accept input from the user
number = int(input("Enter a number: "))

# Call the function and store the result
result = calculate_factorial(number)

# Display the result
if isinstance(result, str):
    print(result)
else:
    print("Factorial:", result)
```

```python
# Assignment 16: Reverse a String
# Write a function named reverse_string().

# Function to reverse a string using slicing
def reverse_string(text):
    return text[::-1]

# Get input from the user
text = input("Enter text: ")

# Call the function
result = reverse_string(text)

# Display the reversed string
print("Reversed text:", result)

# Function to reverse a string using a loop
def reverse_string_loop(text):
    reverse = ""  # Empty string to store the reversed text

    # Traverse each character in the string
    for ch in text:
        reverse = ch + reverse

    # Return the reversed string
    return reverse

# Get input from the user
text = input("Enter text: ")

# Call the function
result = reverse_string_loop(text)

# Display the reversed string
print("Reversed text:", result)
```

```python
# Assignment 17: Palindrome Checker
# Write a function named is_palindrome().

# Function to check whether the text is a palindrome
def is_palindrome(text):
    # Remove spaces and convert to lowercase
    clean_text = text.replace(" ", "").lower()

    # Compare the cleaned text with its reverse
    return clean_text == clean_text[::-1]

# Get input from the user
text = input("Enter text: ")

# Call the function
result = is_palindrome(text)

# Print the result
if result:
    print("The text is a palindrome.")
else:
    print("The text is not a palindrome.")
```

```python
# Assignment 19: Find Maximum in a List
# Write a function named find_list_maximum().

# Function to find the maximum element in a list
def find_list_maximum(numbers):

    # Check if the list is empty
    if len(numbers) == 0:
        return "The list is empty."

    # Assume the first element is the maximum
    maximum = numbers[0]

    # Compare each element with the current maximum
    for num in numbers:
        if num > maximum:
            maximum = num

    # Return the largest element
    return maximum


# Get numbers from the user
user_input = input("Enter numbers: ")

# Check if the user entered anything
if user_input.strip() == "":
    numbers = []
else:
    numbers = list(map(int, user_input.split()))

# Call the function
result = find_list_maximum(numbers)

# Display the result
print("Maximum:", result)
```

```python
# Assignment 20: Count Positive, Negative and Zero Values
# Write a function named count_number_types().

# Function to count positive, negative, and zero values
def count_number_types(numbers):
    positive = 0
    negative = 0
    zero = 0

    # Check each number in the list
    for num in numbers:
        if num > 0:
            positive += 1
        elif num < 0:
            negative += 1
        else:
            zero += 1

    # Return all three counts
    return positive, negative, zero


# Get numbers from the user
numbers = list(map(int, input("Enter numbers: ").split()))

# Receive the returned values using tuple unpacking
positive_count, negative_count, zero_count = count_number_types(numbers)

# Display the results
print("Positive numbers:", positive_count)
print("Negative numbers:", negative_count)
print("Zero values:", zero_count)
```

```python
# Assignment 21: Positional Arguments
# Write a function named employee_details().

# Function to display employee details
def employee_details(name, emp_id, department, salary):
    print("Employee Details")
    print("----------------")
    print("Employee Name :", name)
    print("Employee ID   :", emp_id)
    print("Department    :", department)
    print("Salary        :", salary)
    print()


# First function call using correct positional arguments
employee_details("Ashraf", 101, "IT", 50000)

# Second function call by changing the argument order
employee_details(101, "Ashraf", 50000, "IT")

# Positional arguments are assigned based on their position,
# not by their meaning.
# Changing the order gives incorrect output because the values
# are assigned to the wrong parameters.
```

```python
# Assignment 22: Keyword Arguments
# Use the same employee_details() function.

# Function to display employee details
def employee_details(name, emp_id, department, salary):
    print("Employee Details")
    print("----------------")
    print("Employee Name :", name)
    print("Employee ID   :", emp_id)
    print("Department    :", department)
    print("Salary        :", salary)
    print()


# First function call using keyword arguments
employee_details(
    name="Ashraf",
    emp_id=101,
    department="IT",
    salary=50000
)

# Second function call with the keyword arguments in a different order
employee_details(
    salary=50000,
    department="IT",
    emp_id=101,
    name="Ashraf"
)

# Explanation:
# Keyword arguments assign values using the parameter names.
# Therefore, the order of arguments does not matter.
# Python matches each value with its corresponding parameter name,
# so the output remains correct even when the order is changed.
```

```python
# Assignment 30: Mixed Parameter Types
# Write a function that contains:

# Positional-only parameters
# Normal parameters
# Default parameters
# Variable positional arguments
# Keyword-only parameters
# Variable keyword arguments

# Function demonstrating all parameter types
def generate_report(report_id, /, title, author="Unknown", *sections, status, **details):

    # Positional-only parameter
    print("Report ID      :", report_id)

    # Normal parameter
    print("Title          :", title)

    # Default parameter
    print("Author         :", author)

    # Variable positional arguments
    print("Sections       :")
    for section in sections:
        print(" -", section)

    # Keyword-only parameter
    print("Status         :", status)

    # Variable keyword arguments
    print("Additional Details:")
    for key, value in details.items():
        print(f"{key} : {value}")


# Function call
generate_report(
    101,                        # Positional-only parameter
    "Python Basics",            # Normal parameter
    "Ashraf",                   # Default parameter (overridden)
    "Introduction",             # *args
    "Functions",
    "Arguments",
    status="Completed",         # Keyword-only parameter
    pages=50,                   # **kwargs
    language="English",
    version="1.0"
)
```

```python
# Assignment 31: Return Multiple Values
# Write a function named number_statistics().

# Function to calculate number statistics
def number_statistics(numbers):

    # Calculate statistics using built-in functions
    minimum = min(numbers)
    maximum = max(numbers)
    total = sum(numbers)
    average = total / len(numbers)

    # Return multiple values
    return minimum, maximum, total, average


# Get numbers from the user
numbers = list(map(int, input("Enter numbers: ").split()))

# Receive the returned values using tuple unpacking
minimum, maximum, total, average = number_statistics(numbers)

# Display the results
print("Minimum :", minimum)
print("Maximum :", maximum)
print("Total   :", total)
print("Average :", average)
```

```python
# Assignment 32: Function Returning a Dictionary
# Write a function named analyze_text().

# Function to analyze the given text
def analyze_text(text):

    # Count total characters
    total_characters = len(text)

    # Count total words
    total_words = len(text.split())

    # Count vowels
    vowels = "aeiouAEIOU"
    total_vowels = 0

    for ch in text:
        if ch in vowels:
            total_vowels += 1

    # Count digits
    total_digits = 0
    for ch in text:
        if ch.isdigit():
            total_digits += 1

    # Count spaces
    total_spaces = 0
    for ch in text:
        if ch == " ":
            total_spaces += 1

    # Return all results as a dictionary
    return {
        "Total Characters": total_characters,
        "Total Words": total_words,
        "Total Vowels": total_vowels,
        "Total Digits": total_digits,
        "Total Spaces": total_spaces
    }


# Get input from the user
text = input("Enter text: ")

# Call the function
result = analyze_text(text)

# Display the results
print("\nText Analysis")
print("-------------")

for key, value in result.items():
    print(f"{key}: {value}")
```

```python
# Assignment 33: Function Returning a List
# Write a function named get_even_numbers().

# Function to return all even numbers in a given range
def get_even_numbers(start, end):

    # Validate the range
    if start >= end:
        return "Starting number should be smaller than the ending number."

    # Create an empty list to store even numbers
    even_numbers = []

    # Find even numbers in the given range
    for num in range(start, end + 1):
        if num % 2 == 0:
            even_numbers.append(num)

    # Return the list of even numbers
    return even_numbers


# Get input from the user
start = int(input("Enter starting number: "))
end = int(input("Enter ending number: "))

# Call the function
result = get_even_numbers(start, end)

# Display the result
print("Even Numbers:", result)
```

```python
# Assignment 34: Prime Number Function
# Write a function named is_prime().

# Function to check whether a number is prime
def is_prime(number):

    # Numbers less than 2 are not prime
    if number < 2:
        return False

    # Check divisibility from 2 to number-1
    for i in range(2, number):
        if number % i == 0:
            return False

    # If no divisor is found, the number is prime
    return True


# Print all prime numbers from 1 to 100
print("Prime numbers from 1 to 100:")

for num in range(1, 101):
    if is_prime(num):
        print(num, end=" ")
```

```python
# Assignment 35: Fibonacci Function
# Write a function named generate_fibonacci().

# Function to generate the Fibonacci series
def generate_fibonacci(terms):

    # Handle negative term count
    if terms < 0:
        return "Number of terms cannot be negative."

    # Handle zero terms
    if terms == 0:
        return []

    # Handle one term
    if terms == 1:
        return [0]

    # Initialize the Fibonacci list with the first two numbers
    fibonacci = [0, 1]

    # Generate the remaining terms
    for i in range(2, terms):
        fibonacci.append(fibonacci[i - 1] + fibonacci[i - 2])

    # Return the complete Fibonacci series
    return fibonacci


# Get input from the user
terms = int(input("Enter the number of terms: "))

# Call the function
result = generate_fibonacci(terms)

# Display the result
print("Fibonacci Series:", result)
```

```python
# Assignment 67: Filter Prime Numbers
# Accept a list of integers.

# -----------------------------
# Reusable Prime Function
# -----------------------------

# Function to check whether a number is prime
def is_prime(number):

    # Numbers less than 2 are not prime
    if number < 2:
        return False

    # Check for factors from 2 to number-1
    for i in range(2, number):
        if number % i == 0:
            return False

    return True


# -----------------------------
# User Input
# -----------------------------

# Accept a list of integers from the user
numbers = list(map(int, input("Enter numbers separated by spaces: ").split()))


# -----------------------------
# Filter Prime Numbers
# -----------------------------

# Use filter() with the named function
prime_numbers = list(filter(is_prime, numbers))


# -----------------------------
# Display Results
# -----------------------------

print("\nOriginal List:", numbers)
print("Prime Numbers:", prime_numbers)
```

```python
# Assignment 69: Filter Valid Email Addresses
# Given a list of email strings:

# -----------------------------
# Email Validation Function
# -----------------------------

# Function to check whether an email is valid
def is_valid_email(email):

    # Remove extra spaces
    email = email.strip()

    # Email must contain exactly one '@'
    if email.count("@") != 1:
        return False

    # Split the email into username and domain
    username, domain = email.split("@")

    # Username must not be empty
    if username == "":
        return False

    # Domain must contain at least one dot
    if "." not in domain:
        return False

    # Dot should not be the first or last character of the domain
    if domain.startswith(".") or domain.endswith("."):
        return False

    return True


# -----------------------------
# User Input
# -----------------------------

# Accept email addresses separated by commas
emails = input("Enter email addresses separated by commas: ").split(",")

# -----------------------------
# Filter Valid Emails
# -----------------------------

# Use filter() with the named function
valid_emails = list(filter(is_valid_email, emails))

# -----------------------------
# Display Results
# -----------------------------

print("\nOriginal Emails:")
print(emails)

print("\nValid Emails:")
print(valid_emails)
```

```python
# Assignment 74: Filter Perfect Squares
# Accept a list of positive integers.

# -----------------------------
# Import math Module
# -----------------------------

import math

# -----------------------------
# Perfect Square Function
# -----------------------------

# Function to check whether a number is a perfect square
def is_perfect_square(number):

    # Negative numbers cannot be perfect squares
    if number < 0:
        return False

    # Find the square root
    square_root = math.sqrt(number)

    # Check if the square root is a whole number
    return square_root == int(square_root)


# -----------------------------
# User Input
# -----------------------------

# Accept a list of positive integers
numbers = list(map(int, input("Enter positive integers separated by spaces: ").split()))

# -----------------------------
# Filter Perfect Squares
# -----------------------------

# Use filter() with the named function
perfect_squares = list(filter(is_perfect_square, numbers))

# -----------------------------
# Display Results
# -----------------------------

print("\nOriginal List :", numbers)
print("Perfect Squares:", perfect_squares)
```

```python
# Assignment 81: Calculate Factorial Using reduce()
# Write a function that calculates factorial using reduce().

# -----------------------------
# Import reduce
# -----------------------------

from functools import reduce

# -----------------------------
# Function to Calculate Factorial
# -----------------------------

def calculate_factorial(number):

    # Reject negative numbers
    if number < 0:
        return "Factorial is not defined for negative numbers."

    # Factorial of 0 is 1
    elif number == 0:
        return 1

    # Generate numbers from 1 to the given number
    numbers = range(1, number + 1)

    # Multiply all numbers using reduce()
    factorial = reduce(
        lambda accumulator, value: accumulator * value,
        numbers
    )

    return factorial


# -----------------------------
# User Input
# -----------------------------

number = int(input("Enter a non-negative integer: "))

# -----------------------------
# Display Result
# -----------------------------

result = calculate_factorial(number)

print("\nFactorial:", result)
```

```python
# Assignment 83: Find the Longest Word Using reduce()
# Accept a sentence.

# -----------------------------
# Import reduce
# -----------------------------

from functools import reduce

# -----------------------------
# User Input
# -----------------------------

# Accept a sentence from the user
sentence = input("Enter a sentence: ")

# Handle empty input
if sentence.strip() == "":
    words = []
else:
    # Split the sentence into words
    words = sentence.split()

# -----------------------------
# Find the Longest Word
# -----------------------------

if len(words) == 0:
    print("\nNo words found.")
else:
    # Compare two words and keep the longer one.
    # If both words have the same length,
    # the first word is kept.
    longest_word = reduce(
        lambda first, second: first if len(first) >= len(second) else second,
        words
    )

    # -----------------------------
    # Display Result
    # -----------------------------

    print("\nWords:", words)
    print("Longest Word:", longest_word)
```

```python
# Assignment 91: Closure-Based Counter
# Write a function named create_counter().

# -----------------------------
# Outer Function
# -----------------------------

def create_counter():

    # Initialize the counter
    count = 0

    # -----------------------------
    # Inner Function
    # -----------------------------

    def counter():

        # Use nonlocal to modify the outer variable
        nonlocal count

        # Increase the count
        count += 1

        # Return the updated count
        return count

    # Return the inner function
    return counter


# -----------------------------
# Create Counter
# -----------------------------

counter = create_counter()

# -----------------------------
# Call the Counter Function
# -----------------------------

print("Count:", counter())
print("Count:", counter())
print("Count:", counter())
print("Count:", counter())
print("Count:", counter())
```

```python
# Assignment 92: Closure-Based Bank Balance
# Write a function named create_bank_account().

# -----------------------------
# Outer Function
# -----------------------------

def create_bank_account(initial_balance):

    # Store the balance
    balance = initial_balance

    # -----------------------------
    # Deposit Function
    # -----------------------------

    def deposit(amount):
        nonlocal balance

        balance += amount
        print("Deposited:", amount)

    # -----------------------------
    # Withdraw Function
    # -----------------------------

    def withdraw(amount):
        nonlocal balance

        if amount > balance:
            print("Insufficient Balance!")
        else:
            balance -= amount
            print("Withdrawn:", amount)

    # -----------------------------
    # Balance Enquiry Function
    # -----------------------------

    def check_balance():
        print("Current Balance:", balance)

    # Return all three functions
    return deposit, withdraw, check_balance


# -----------------------------
# Create Bank Account
# -----------------------------

deposit, withdraw, check_balance = create_bank_account(10000)

# -----------------------------
# Perform Operations
# -----------------------------

check_balance()

deposit(2500)
check_balance()

withdraw(4000)
check_balance()

withdraw(10000)
check_balance()
```

```python
# Assignment 93: Function Decorator
# Create a decorator named execution_logger().

# -----------------------------
# Decorator Function
# -----------------------------

def execution_logger(function):

    # Wrapper function
    def wrapper(num1, num2):

        # Message before function execution
        print("Starting function...")

        # Call the original function
        result = function(num1, num2)

        # Print the returned result
        print("Result:", result)

        # Message after function execution
        print("Function completed.")

        # Return the original function's result
        return result

    return wrapper


# -----------------------------
# Addition Function
# -----------------------------

@execution_logger
def add(a, b):
    return a + b


# -----------------------------
# User Input
# -----------------------------

number1 = int(input("Enter first number: "))
number2 = int(input("Enter second number: "))

# -----------------------------
# Call the Function
# -----------------------------

add(number1, number2)
```

```python
# Assignment 99: LEGB Name Resolution
# Create variables with the same name at multiple scope levels.

# -----------------------------
# Global Scope
# -----------------------------

# Global variable
value = "Global Value"

# -----------------------------
# Outer Function
# -----------------------------

def outer_function():

    # Enclosing variable
    value = "Enclosing Value"

    print("From Outer Function :", value)

    # -----------------------------
    # Inner Function
    # -----------------------------

    def inner_function():

        # Local variable
        value = "Local Value"

        print("From Inner Function :", value)

    # Call the inner function
    inner_function()


# -----------------------------
# Main Program
# -----------------------------

# Print the global value
print("From Global Scope :", value)

# Call the outer function
outer_function()


# ----------------------------------------------------
# Experiment 1:
# Remove the line:
#
# value = "Local Value"
#
# from inner_function().
#
# Then the output inside inner_function() becomes:
# "Enclosing Value"
#
# Python cannot find 'value' in the Local scope,
# so it looks in the Enclosing scope.
# ----------------------------------------------------


# ----------------------------------------------------
# Experiment 2:
# Also remove the line:
#
# value = "Enclosing Value"
#
# from outer_function().
#
# Now the output inside inner_function() becomes:
# "Global Value"
#
# Python cannot find 'value' in the Local scope,
# then not in the Enclosing scope,
# so it searches the Global scope.
# ----------------------------------------------------


# ----------------------------------------------------
# LEGB Rule
# ----------------------------------------------------
#
# Python searches for a variable in this order:
#
# L → Local
# E → Enclosing
# G → Global
# B → Built-in
#
# 1. Local
#    Variables inside the current function.
#
# 2. Enclosing
#    Variables in the outer (enclosing) function.
#
# 3. Global
#    Variables declared outside all functions.
#
# 4. Built-in
#    Python's predefined names like:
#    print(), len(), max(), input(), etc.
#
# Python stops searching as soon as it finds the variable.
```

```python
# Part 5: Salary Increase
# Use map() to calculate revised salaries based on these conditions:

# Performance score 90 or above: 15% increase
# Performance score 80–89: 10% increase
# Performance score 70–79: 5% increase
# Below 70: No increase
# Do not modify the original employee dictionaries directly.

# -----------------------------
# Function to Calculate Revised Salaries
# -----------------------------

def calculate_revised_salary(employee):

    # Create a copy of the employee dictionary
    # so the original data is not modified
    updated_employee = employee.copy()

    # Get the performance score
    score = employee["performance_score"]

    # Calculate revised salary based on performance
    if score >= 90:
        updated_employee["salary"] = employee["salary"] * 1.15
    elif score >= 80:
        updated_employee["salary"] = employee["salary"] * 1.10
    elif score >= 70:
        updated_employee["salary"] = employee["salary"] * 1.05
    else:
        updated_employee["salary"] = employee["salary"]

    # Return the updated employee dictionary
    return updated_employee


# -----------------------------
# Apply Salary Revision Using map()
# -----------------------------

revised_employees = list(map(calculate_revised_salary, employees))

# -----------------------------
# Display Results
# -----------------------------

print("Original Salary -> Revised Salary\n")

for original, revised in zip(employees, revised_employees):
    print(f"{original['name']}: ₹{original['salary']} -> ₹{revised['salary']}")
```

```python
# Part 6: Total Salary Expense
# Use reduce() to calculate:

# Current total salary
# Revised total salary
# Total additional salary expense

from functools import reduce

# -----------------------------
# Calculate Current Total Salary
# -----------------------------

current_total_salary = reduce(
    lambda total, employee: total + employee["salary"],
    employees,
    0
)

# -----------------------------
# Calculate Revised Total Salary
# -----------------------------

revised_total_salary = reduce(
    lambda total, employee: total + employee["salary"],
    revised_employees,
    0
)

# -----------------------------
# Calculate Additional Salary Expense
# -----------------------------

additional_salary_expense = revised_total_salary - current_total_salary

# -----------------------------
# Display Results
# -----------------------------

print("Current Total Salary      : ₹", current_total_salary)
print("Revised Total Salary      : ₹", revised_total_salary)
print("Additional Salary Expense : ₹", additional_salary_expense)
```

```python
# Part 7: Highest-Paid Employee
# Use reduce() to find the employee with the highest salary.

# Do not use max().

from functools import reduce

# -----------------------------
# Find the Highest-Paid Employee
# -----------------------------

highest_paid_employee = reduce(
    lambda emp1, emp2: emp1 if emp1["salary"] > emp2["salary"] else emp2,
    employees
)

# -----------------------------
# Display Result
# -----------------------------

print("Highest-Paid Employee:\n")
print(highest_paid_employee)
```

```python
# Part 9: Closure-Based Bonus Calculator
# Create a function named create_bonus_calculator().

# The outer function should accept a bonus percentage.
# The inner function should accept a salary.
# Return the calculated bonus.
# Create separate bonus calculators for 5%, 10%, and 15%.

# -----------------------------
# Closure-Based Bonus Calculator
# -----------------------------

def create_bonus_calculator(bonus_percentage):

    # Inner function
    def calculate_bonus(salary):
        # Calculate and return the bonus amount
        return salary * bonus_percentage / 100

    # Return the inner function
    return calculate_bonus


# -----------------------------
# Create Bonus Calculators
# -----------------------------

bonus_5 = create_bonus_calculator(5)
bonus_10 = create_bonus_calculator(10)
bonus_15 = create_bonus_calculator(15)

# -----------------------------
# Display Bonus for Each Employee
# -----------------------------

for employee in employees:
    print(f"\nEmployee: {employee['name']}")
    print(f"Salary   : ₹{employee['salary']}")
    print(f"5% Bonus : ₹{bonus_5(employee['salary'])}")
    print(f"10% Bonus: ₹{bonus_10(employee['salary'])}")
    print(f"15% Bonus: ₹{bonus_15(employee['salary'])}")
```

```python
# Part 10: Nested Report Generator
# Create an outer function named generate_department_report().

# Inside it, define nested functions to calculate:

# Number of employees
# Total salary
# Average salary
# Average performance score
# Return the complete department report as a dictionary.

# -----------------------------
# Nested Department Report Generator
# -----------------------------

def generate_department_report(department):

    # Filter employees belonging to the given department
    department_employees = list(
        filter(lambda employee: employee["department"].lower() == department.lower(), employees)
    )

    # Nested function to count employees
    def employee_count():
        return len(department_employees)

    # Nested function to calculate total salary
    def total_salary():
        return sum(employee["salary"] for employee in department_employees)

    # Nested function to calculate average salary
    def average_salary():
        if employee_count() == 0:
            return 0
        return total_salary() / employee_count()

    # Nested function to calculate average performance score
    def average_performance():
        if employee_count() == 0:
            return 0
        return sum(employee["performance_score"] for employee in department_employees) / employee_count()

    # Return the department report as a dictionary
    return {
        "Department": department,
        "Number of Employees": employee_count(),
        "Total Salary": total_salary(),
        "Average Salary": average_salary(),
        "Average Performance Score": average_performance()
    }


# -----------------------------
# Example
# -----------------------------

department_name = input("Enter Department Name: ")

report = generate_department_report(department_name)

print("\nDepartment Report:")
for key, value in report.items():
    print(f"{key}: {value}")
```

```python
# Part 11: LEGB Demonstration
# Create a global variable:

# company_name = "Edukron"
# Inside the report function:

# Access the global company name.
# Create an enclosing variable for the department.
# Create a local variable for the report title.
# Add comments explaining how Python resolves each variable using LEGB.

# -----------------------------
# Global Variable
# -----------------------------

company_name = "Edukron"

# -----------------------------
# Report Function
# -----------------------------

def generate_department_report(department):

    # Enclosing variable
    department_name = department

    def generate_report():

        # Local variable
        report_title = "Department Report"

        # Local Scope
        print("Report Title :", report_title)

        # Enclosing Scope
        print("Department   :", department_name)

        # Global Scope
        print("Company      :", company_name)

    # Call the inner function
    generate_report()


# -----------------------------
# Main Program
# -----------------------------

department = input("Enter Department Name: ")

generate_department_report(department)

# ----------------------------------------------------
# LEGB Rule Explanation
# ----------------------------------------------------
# L → Local
#     report_title is found inside generate_report().
#
# E → Enclosing
#     department_name is found in generate_department_report().
#
# G → Global
#     company_name is declared outside all functions.
#
# B → Built-in
#     Python searches built-in names like print(), len(), input()
#     only if the variable is not found in Local, Enclosing,
#     or Global scope.
#
# Python always searches variables in this order:
# Local → Enclosing → Global → Built-in
```
