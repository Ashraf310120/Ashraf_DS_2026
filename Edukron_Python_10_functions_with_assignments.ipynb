```python
# Assignment 3: Calculator Using Functions
# Create four separate functions:

# add()
# subtract()
# multiply()
# divide()

# Function to add two numbers
def add(num1, num2):
    return (num1 + num2)

# Function to subtract two numbers
def subtract(num1, num2):
    return (num1 - num2)

# Function to multiply two numbers
def multiply(num1, num2):
    return (num1 * num2)

# Function to divide two numbers
def divide(num1, num2):
    # Check for division by zero
    if num2 == 0:
        return "Error! Division by zero is not allowed."
    return num1 / num2


# Accept input from the user
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))
operation = input("Enter operation (add, subtract, multiply, divide): ").lower()

# Call the appropriate function based on the user's choice
if operation == "add":
    result = add(num1, num2)

elif operation == "subtract":
    result = subtract(num1, num2)

elif operation == "multiply":
    result = multiply(num1, num2)

elif operation == "divide":
    result = divide(num1, num2)

else:
    result = "Invalid operation!"

# Display the result
print("Result:", result)
```

```python
# Assignment 5: Find the Largest Number
# Write a function named find_largest() that accepts three numbers.

# Function definition
# This function finds the largest of three numbers.
def find_largest(num1, num2, num3):
    # Assume the first number is the largest
    largest = num1

    # Compare with the second number
    if num2 > largest:
        largest = num2

    # Compare with the third number
    if num3 > largest:
        largest = num3

    # Return the largest number
    return largest


# Accept three numbers from the user
num1, num2, num3 = map(int, input("Enter three numbers: ").split())

# Call the function and store the result
result = find_largest(num1, num2, num3)

# Display the result
print("Largest number:", result)
```

```python
# Assignment 7: Area of a Rectangle
# Write a function named rectangle_area().

# Function definition
# This function calculates the area of a rectangle.
def rectangle_area(length, width):
    # Check if the length and width are positive
    if length <= 0 or width <= 0:
        return "Error! Length and width must be positive."

    # Calculate and return the area
    return length * width


# Accept input from the user
length = float(input("Enter length: "))
width = float(input("Enter width: "))

# Call the function and store the result
result = rectangle_area(length, width)

# Display the result
if isinstance(result, str):
    print(result)
else:
    print("Area of rectangle:", result)
```

```python
# Assignment 8: Area and Circumference of a Circle
# Create two functions:

# circle_area(radius)
# circle_circumference(radius)

# Import the math module
import math

# Function definition
# This function calculates the area of a circle.
def circle_area(radius):
    return math.pi * radius ** 2

# Function definition
# This function calculates the circumference of a circle.
def circle_circumference(radius):
    return 2 * math.pi * radius

# Accept input from the user
radius = float(input("Enter radius: "))

# Call the functions and store the results
area = circle_area(radius)
circumference = circle_circumference(radius)

# Display the results rounded to two decimal places
print("Area:", round(area, 2))
print("Circumference:", round(circumference, 2))
```

```python
# Assignment 9: Student Grade Calculator
# Write a function named calculate_grade().

# Function definition
# This function calculates the grade based on the student's marks.
def calculate_grade(marks):
    # Validate the marks
    if marks < 0 or marks > 100:
        return "Error! Marks should be between 0 and 100."

    # Determine the grade
    if marks >= 90:
        return "A"
    elif marks >= 80:
        return "B"
    elif marks >= 70:
        return "C"
    elif marks >= 60:
        return "D"
    else:
        return "F"


# Accept marks from the user
marks = int(input("Enter marks: "))

# Call the function and store the result
result = calculate_grade(marks)

# Display the result
if result.startswith("Error"):
    print(result)
else:
    print("Grade:", result)
```

```python
# Assignment 10: Simple Interest Calculator
# Write a function named simple_interest().

# Function definition
# This function calculates the simple interest.
def simple_interest(principal, rate, time):
    # Calculate and return the simple interest
    return (principal * rate * time) / 100


# Accept input from the user
principal = float(input("Principal: "))
rate = float(input("Rate: "))
time = float(input("Time: "))

# Call the function and store the result
interest = simple_interest(principal, rate, time)

# Display the result with two decimal places
print("Simple Interest:", format(interest, ".2f"))
```

```python
# Assignment 11: Celsius to Fahrenheit
# Write a function named celsius_to_fahrenheit().

# Function definition
# This function converts Celsius to Fahrenheit.
def celsius_to_fahrenheit(celsius):
    # Convert Celsius to Fahrenheit and return the result
    return (celsius * 9 / 5) + 32


# Accept temperature in Celsius from the user
celsius = float(input("Enter temperature in Celsius: "))

# Call the function and store the result
fahrenheit = celsius_to_fahrenheit(celsius)

# Display the result rounded to two decimal places
print("Temperature in Fahrenheit:", round(fahrenheit, 2))
```

```python
# Assignment 12: Check Leap Year
# Write a function named is_leap_year().

# Function definition
# This function checks whether a given year is a leap year.
def is_leap_year(year):
    # Check the leap year conditions
    if year % 400 == 0:
        return True
    elif year % 4 == 0 and year % 100 != 0:
        return True
    else:
        return False


# Accept the year from the user
year = int(input("Enter a year: "))

# Call the function and store the result
result = is_leap_year(year)

# Display the appropriate message
if result:
    print(year, "is a Leap Year.")
else:
    print(year, "is not a Leap Year.")
```

```python
# Assignment 13: Factorial Using a Function
# Write a function named calculate_factorial().

# Function definition
# This function calculates the factorial of a non-negative integer.
def calculate_factorial(number):
    # Check for negative input
    if number < 0:
        return "Error! Factorial is not defined for negative numbers."

    # Initialize factorial to 1
    factorial = 1

    # Calculate the factorial using a loop
    for i in range(1, number + 1):
        factorial *= i

    # Return the factorial
    return factorial


# Accept input from the user
number = int(input("Enter a number: "))

# Call the function and store the result
result = calculate_factorial(number)

# Display the result
if isinstance(result, str):
    print(result)
else:
    print("Factorial:", result)
```

```python
# Assignment 14: Multiplication Table Function
# Write a function named multiplication_table().

# Function definition
# This function prints the multiplication table of a number.
# The default limit is 10.
def multiplication_table(number, limit=10):
    # Print the multiplication table
    for i in range(1, limit + 1):
        print(number, "×", i, "=", number * i)


# Accept a number from the user
number = int(input("Enter a number: "))

# Call the function with the default limit
print("\nMultiplication Table (Default Limit):")
multiplication_table(number)

# Accept a custom limit from the user
custom_limit = int(input("\nEnter a custom limit: "))

# Call the function with the custom limit
print("\nMultiplication Table (Custom Limit):")
multiplication_table(number, custom_limit)
```

```python
# Assignment 15: Count Vowels in a String
# Write a function named count_vowels().

def count_vowels(text):
    vowels = "aeiouAEIOU"
    count = 0

    for ch in text:
        if ch in vowels:
            count += 1

    return count


text = input("Enter text: ")
result = count_vowels(text)

print("Number of vowels:", result)
```

```python
# Assignment 16: Reverse a String
# Write a function named reverse_string().

# Function to reverse a string using slicing
def reverse_string(text):
    return text[::-1]

# Get input from the user
text = input("Enter text: ")

# Call the function
result = reverse_string(text)

# Display the reversed string
print("Reversed text:", result)

# Function to reverse a string using a loop
def reverse_string_loop(text):
    reverse = ""  # Empty string to store the reversed text

    # Traverse each character in the string
    for ch in text:
        reverse = ch + reverse

    # Return the reversed string
    return reverse

# Get input from the user
text = input("Enter text: ")

# Call the function
result = reverse_string_loop(text)

# Display the reversed string
print("Reversed text:", result)
```

```python
# Assignment 17: Palindrome Checker
# Write a function named is_palindrome().

# Function to check whether the text is a palindrome
def is_palindrome(text):
    # Remove spaces and convert to lowercase
    clean_text = text.replace(" ", "").lower()

    # Compare the cleaned text with its reverse
    return clean_text == clean_text[::-1]

# Get input from the user
text = input("Enter text: ")

# Call the function
result = is_palindrome(text)

# Print the result
if result:
    print("The text is a palindrome.")
else:
    print("The text is not a palindrome.")
```

```python
# Assignment 18: Sum of List Elements
# Write a function named calculate_list_sum().


# Function to calculate the sum of list elements
def calculate_list_sum(numbers):
    total = 0

    # Add each number using a loop
    for num in numbers:
        total += num

    # Return the total sum
    return total


# Get numbers from the user
numbers = list(map(int, input("Enter numbers: ").split()))

# Call the function
result = calculate_list_sum(numbers)

# Display the total
print("Total:", result)
```

```python
# Assignment 19: Find Maximum in a List
# Write a function named find_list_maximum().

# Function to find the maximum element in a list
def find_list_maximum(numbers):

    # Check if the list is empty
    if len(numbers) == 0:
        return "The list is empty."

    # Assume the first element is the maximum
    maximum = numbers[0]

    # Compare each element with the current maximum
    for num in numbers:
        if num > maximum:
            maximum = num

    # Return the largest element
    return maximum


# Get numbers from the user
user_input = input("Enter numbers: ")

# Check if the user entered anything
if user_input.strip() == "":
    numbers = []
else:
    numbers = list(map(int, user_input.split()))

# Call the function
result = find_list_maximum(numbers)

# Display the result
print("Maximum:", result)
```

```python
# Assignment 20: Count Positive, Negative and Zero Values
# Write a function named count_number_types().

# Function to count positive, negative, and zero values
def count_number_types(numbers):
    positive = 0
    negative = 0
    zero = 0

    # Check each number in the list
    for num in numbers:
        if num > 0:
            positive += 1
        elif num < 0:
            negative += 1
        else:
            zero += 1

    # Return all three counts
    return positive, negative, zero


# Get numbers from the user
numbers = list(map(int, input("Enter numbers: ").split()))

# Receive the returned values using tuple unpacking
positive_count, negative_count, zero_count = count_number_types(numbers)

# Display the results
print("Positive numbers:", positive_count)
print("Negative numbers:", negative_count)
print("Zero values:", zero_count)
```
